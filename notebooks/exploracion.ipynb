# Exploración de AG News
Cifras agregadas, sin imprimir ni redistribuir noticias. Ejecutar desde la raíz del repositorio.

```python
from collections import Counter
from datasets import load_dataset
import matplotlib.pyplot as plt

train = load_dataset('fancyzhx/ag_news', split='train')
names = train.features['label'].names
counts = Counter(train['label'])
print({names[i]: counts[i] for i in range(len(names))})
plt.figure(figsize=(7, 4)); plt.bar(names, [counts[i] for i in range(len(names))]); plt.ylabel('Noticias en entrenamiento');
```

```python
lengths = [len(text.split()) for text in train['text']]
print('Media (palabras):', round(sum(lengths) / len(lengths), 2))
plt.figure(figsize=(7, 4)); plt.hist(lengths, bins=40); plt.xlabel('Palabras por noticia'); plt.ylabel('Frecuencia');
```
